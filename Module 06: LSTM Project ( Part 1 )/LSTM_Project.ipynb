# **📖 Language Modeling Concept (English Version)**

Language modeling is the core concept behind text generation. It is the exact mechanism that advanced AI models like ChatGPT use to predict the next word in a sequence to generate human-like text.

Let's look at an example sentence: **"The Last Train Journey"**

If we want to build a model to generate this text, we must frame it as a **Supervised Learning** problem. But how do we define the Features (input) and the Target (output)? We do this by using the previous words to predict the next word:

| Features (X) | Target (y) |
| --- | --- |
| The | Last |
| The Last | Train |
| The Last Train | Journey |

By structuring the data this way, we successfully separate our features and targets. This next-word prediction process is the fundamental concept of Language Modeling.

### 🛠️ Project Steps

To build our text generation project, we will follow these 4 steps:

1. **Form Vocabulary:** Create a dictionary of all the unique words present in our dataset.
2. **Convert Text to Indices:** Since machines only understand numbers, we will convert each word into a unique numerical index.
3. **Form Sequences:** We will structure our features and targets using these numbers.
| Features (Indices) | Target (Index) |
| --- | --- |
| The [1] | Last [2] |
| The [1], Last [2] | Train [3] |
| The [1], Last [2], Train [3] | Journey [4] |


4. **Word Embedding:** Finally, we will convert each numerical index into a dense vector (e.g., a 50-dimensional vector). This allows the machine to understand the semantic meaning and relationships between words. Common techniques discussed in NLP include Word2Vec, GloVe, TF-IDF, or standard Neural Network Embedding layers.

# **Form vocabulary**

### 📖 Why do we build a Vocabulary?
Deep learning models (like LSTMs) are essentially massive calculators. They cannot read or understand raw text (like "train" or "journey"). They only understand numbers (matrices and tensors).

Therefore, we need to create a **"Dictionary" or "Lookup Table"** that maps every unique word in our dataset to a unique integer (number). This mapping is called a Vocabulary.

**Why the `<unk>` token?**
When our model is deployed in the real world, a user might input a word that the model has never seen during training (e.g., "helicopter"). If the model doesn't have an ID for that word, it will crash. To prevent this, we create an `<unk>` (Unknown) token. Any word not found in our vocabulary is automatically assigned the ID of `<unk>` (which is 0), allowing the model to process it without crashing.

* **What it does:** This is a safety check. It ensures we don't accidentally overwrite an existing token (like if `'<unk>'` happened to be in the text).

```python
# Assign a new, unique numerical ID to the word.
    vocab[token] = len(vocab)

```

* **What it does:** This is the most clever part of the code! It assigns the *current length* of the dictionary as the new ID.
* Initially, `vocab` has 1 item (`<unk>`). So, the first new word gets ID `1`.
* Now the dictionary has 2 items. The next word gets ID `2`.
* This automatically creates sequential numbers (1, 2, 3, 4...) for every new word.

### 📖 Vocabulary কেন তৈরি করা হয়? (বাংলা ভার্সন)

ডিপ লার্নিং মডেল (যেমন LSTM) মূলত একটি বিশাল ক্যালকুলেটর। এরা সরাসরি কোনো টেক্সট বা শব্দ (যেমন "train" বা "journey") পড়তে বা বুঝতে পারে না; এরা শুধু নাম্বার (ম্যাট্রিক্স এবং টেন্সর) বোঝে।

এই কারণে, টেক্সটকে নাম্বারে কনভার্ট করার জন্য আমাদের একটি **"ডিকশনারি" বা "লুকআপ টেবিল (Lookup Table)"** তৈরি করতে হয়, যেখানে আমাদের ডেটাসেটের প্রতিটি ইউনিক শব্দের বিপরীতে একটি নির্দিষ্ট নাম্বার (ID) দেওয়া থাকে। এই ম্যাপিং বা ডিকশনারিটিকেই Vocabulary বলা হয়।

**`<unk>` টোকেন কেন লাগে?**
মডেলটি যখন বাস্তবে (Real-world) কাজ করবে, তখন ইউজার এমন কোনো শব্দ ইনপুট দিতে পারে যা মডেলটি আগে কখনো দেখেনি (যেমন "helicopter")। সেই শব্দের যদি কোনো নাম্বার বা ID না থাকে, তাহলে কোড ক্র্যাশ করবে। এই সমস্যা সমাধানের জন্যই `<unk>` (Unknown) টোকেন ব্যবহার করা হয়। মডেল যখনই কোনো অপরিচিত শব্দ পায়, তখন সে সেটিকে `<unk>`-এর ID (অর্থাৎ 0) দিয়ে রিপ্লেস করে দেয়। ফলে মডেল ক্র্যাশ না করে কাজ চালিয়ে যেতে পারে।

### Extracting Sentences and Tokenization (English Version)

Our main task begins here. First, we need to extract individual sentences from our raw document, clean them, and then tokenize them into numerical formats. Without this, our model cannot process the text.

#### 1. Extracting and Cleaning Sentences

If we print our raw `document`, we will see newline characters (`\n`). If we just use `document.split('\n')`, we get a list of sentences, but it also includes empty strings (`''`). We need to clean this up.

```python
# Splitting by newline and filtering out empty lines in one go
input_sentences = [line for line in document.split('\n') if line.strip() != '']

input_sentences

```

**Line-by-Line Explanation:**

* `document.split('\n')`: Breaks the entire story into a list of separate lines wherever there is a line break.
* `for line in ...`: Loops through that newly generated list of lines.
* `if line.strip() != ''`: Checks if a line is empty. If it has actual text, it keeps it; if it is just an empty space, it throws it away.
* `input_sentences`: Now holds our perfectly clean list of sentences (e.g., `['The Last Train Journey', 'On a cold winter evening...', ...]`).

#### 2. Tokenizing and Converting to Numerical Indices

Now that we have clean sentences, we need to convert every word in these sentences into its corresponding ID from our `vocab` dictionary.

```python
# Function to convert a single sentence into a list of numbers
def text_to_indices(sentence, vocab):
    numerical_sentence = []
    for token in sentence:
        if token in vocab:
            numerical_sentence.append(vocab[token])
        else:
            numerical_sentence.append(vocab['<unk>'])
    return numerical_sentence

# List to store all numerical sentences
input_numerical_sentecnes = []

# Loop through our cleaned sentences
for sentence in input_sentences:
    input_numerical_sentecnes.append(text_to_indices(word_tokenize(sentence.lower()), vocab))

```

**Line-by-Line Explanation:**

* `def text_to_indices(sentence, vocab):`: Defines a custom function that takes a tokenized sentence and our vocabulary dictionary as inputs.
* `numerical_sentence = []`: Creates an empty list to store the numerical IDs for the current sentence.
* `for token in sentence:`: Loops through every word (token) in the given sentence.
* `if token in vocab:`: Checks if the word exists in our vocabulary.
* `numerical_sentence.append(vocab[token])`: If it exists, it finds the ID and adds it to our list.
* `else: numerical_sentence.append(vocab['<unk>'])`: If the word is missing, it adds the ID for the unknown token (`0`) to prevent errors.
* `input_numerical_sentecnes = []`: Creates the master list to hold all the finalized numerical sentences.
* `for sentence in input_sentences:`: Loops through our clean, text-based sentences.
* `word_tokenize(sentence.lower())`: Converts the sentence to lowercase and splits it into individual words using NLTK.
* `text_to_indices(...)`: Passes those words into our custom function to get numbers, and then appends that list of numbers to `input_numerical_sentecnes`.

# **📝 Forming Training Sequences**

Now we need to form our training sequences. For a supervised learning model to predict the next word, it needs to see how a sentence builds up word by word.

If our numerical sentence is `[1, 2, 3, 4]`, we cannot just feed the whole array to the model at once. We must break it down into smaller, growing sequences like this:

* `[1, 2]` (Model sees 1, predicts 2)
* `[1, 2, 3]` (Model sees 1 and 2, predicts 3)
* `[1, 2, 3, 4]` (Model sees 1, 2, and 3, predicts 4)

**The Final Code:**

```python
# Create an empty list to store all our generated sequences
training_sequence = []

# Loop through each numerical sentence
for sentence in input_numerical_sentecnes:
    # Loop through the indices of the sentence, starting from index 1
    for i in range(1, len(sentence)):
        # Slice the sentence from the start up to the current index + 1
        training_sequence.append(sentence[:i+1])

# Check the generated sequences
training_sequence

```

**Line-by-Line Explanation:**

* `training_sequence = []`: Creates an empty master list where all our newly sliced sequences will be stored.
* `for sentence in input_numerical_sentecnes:`: This outer loop goes through every single full numerical sentence in our dataset.
* `for i in range(1, len(sentence)):`: This inner loop controls the slicing. Notice that it starts at `1`, not `0`. Why? Because a sequence needs at least two words (one as input/feature, one as target). If we started at `0`, our slice would just be `[1]`, which has no target word to predict.
* `sentence[:i+1]`: This is Python list slicing. It grabs all numbers from the beginning of the list (`0`) up to the position `i`. As `i` increases, the slice gets longer.
* `training_sequence.append(...)`: Adds that freshly sliced piece of the sentence into our master list.

# **Padding training sequence**

#### ⚠️ The Problem: Variable Sequence Lengths

If we look at the first 5 sequences we generated:
`[[1, 2], [1, 2, 3], [1, 2, 3, 4], [5, 6], [5, 6, 7]]`

We immediately see a major issue: **the lengths of these lists are not the same.** Neural networks (like LSTMs) process data in "batches" (e.g., 32 sequences at a time) to train efficiently. A batch is essentially a matrix, and a matrix must have perfectly equal rows and columns. If our sequences have different lengths, the model will throw an error because it cannot form a uniform batch.

#### 💡 The Solution: Pre-Padding with Zeros

To fix this, we need to make all sequences as long as the longest sequence in our dataset. We do this by adding `0`s to the beginning of the shorter sequences. This technique is called **Pre-padding**. The model ignores these `0`s (since `0` is our `<unk>` token, or acts as neutral padding), but it allows the data to fit perfectly into a matrix.

**The Code:**

```python
# Step A: Find the maximum sequence length
len_list = []
for sequence in training_sequence:
    len_list.append(len(sequence))

max_len = max(len_list)

# Step B: Apply padding to all sequences
padded_sequence = []
for sequence in training_sequence:
    # Generate the required number of zeros and add them to the front
    padded_sequence.append([0] * (max_len - len(sequence)) + sequence)

# Check the final result
padded_sequence

```

**Line-by-Line Explanation:**

* `len_list = []`: Creates an empty list to store the lengths of all our sequences.
* `len_list.append(len(sequence))`: Calculates the length of each sequence and adds it to our list.
* `max_len = max(len_list)`: Finds the largest number in that list. This is the length of our longest sentence.
* `padded_sequence = []`: Creates a master list for our final, equal-sized sequences.
* `[0] * (max_len - len(sequence))`: This is brilliant Python logic! It subtracts the current sequence's length from the `max_len` to find out how many zeros are missing. Then, it multiplies a list containing zero `[0]` by that number to generate exactly that many zeros (e.g., `[0] * 3` becomes `[0, 0, 0]`).
* `... + sequence`: Concatenates (joins) those generated zeros to the *front* of the original sequence.

### 🛠️ ট্রেনিং সিকোয়েন্স প্যাডিং করা (বাংলা ভার্সন)

#### ⚠️ সমস্যা: সিকোয়েন্সের ভিন্ন সাইজ বা দৈর্ঘ্য

আমরা যদি আমাদের তৈরি করা প্রথম ৫টি সিকোয়েন্সের দিকে তাকাই:
`[[1, 2], [1, 2, 3], [1, 2, 3, 4], [5, 6], [5, 6, 7]]`

আমরা খুব সহজেই একটি বড় সমস্যা দেখতে পাব: **লিস্টগুলোর সাইজ বা দৈর্ঘ্য সমান নয়।**
নিউরাল নেটওয়ার্ক (যেমন LSTM) দ্রুত কাজ করার জন্য ডেটাকে "ব্যাচ" (Batch) আকারে প্রসেস করে (যেমন: একসাথে ৩২টি সিকোয়েন্স পাঠানো)। একটি ব্যাচ মূলত একটি ম্যাট্রিক্স, যার সারি (row) এবং কলাম (column) একদম সমান হতে হয়। আমাদের সিকোয়েন্সগুলো ছোট-বড় হলে মডেল কোনো ম্যাট্রিক্স তৈরি করতে পারবে না এবং এরর (error) দেবে।

#### 💡 সমাধান: শুরুতে শূন্য (0) যোগ করা (Pre-Padding)

এই সমস্যার সমাধানের জন্য, আমাদের ডেটাসেটের সবচেয়ে বড় সিকোয়েন্সটি খুঁজে বের করতে হবে এবং বাকি ছোট সিকোয়েন্সগুলোর শুরুতে প্রয়োজনীয় সংখ্যক `0` (শূন্য) বসিয়ে সবগুলোকে সমান সাইজে নিয়ে আসতে হবে। একে বলা হয় **Pre-padding**। মডেল এই শূন্যগুলোকে ইগনোর করে, কিন্তু এর ফলে ডেটাগুলো সুন্দরভাবে ম্যাট্রিক্সে সেট হয়ে যায়।

* `... + sequence`: তৈরি করা শূন্যগুলোর সাথে অরিজিনাল সিকোয়েন্সটিকে যুক্ত করে দেয়। যেহেতু শূন্যগুলো আগে আছে, তাই এটি সিকোয়েন্সের শুরুতেই বসে যায়।

# **Convert to Tensor and Covert X and y**

Before we can feed our perfectly padded sequences into a PyTorch model, we must convert our standard Python lists into **PyTorch Tensors**. After the conversion, we will apply the slicing logic we just learned to separate our inputs ($X$) from our targets ($y$).

**The Code:**

```python
# Convert the Python list into a PyTorch Tensor
padded_training_sequence = torch.tensor(padded_sequence, dtype=torch.long)

# Verify the tensor
padded_training_sequence

# Extract Features (X) and Targets (y)
X = padded_training_sequence[:, :-1]
y = padded_training_sequence[:, -1]

```

**Line-by-Line Explanation:**

* `torch.tensor(padded_sequence, ...)`: This function takes our standard Python list (`padded_sequence`) and converts it into a PyTorch tensor, which is the specialized matrix format required for deep learning operations.
* `dtype=torch.long`: This is a crucial step! It forces the numbers to be 64-bit integers. Since our data consists of vocabulary IDs (whole numbers like 1, 2, 21), PyTorch's `Embedding` layer and `CrossEntropyLoss` function strictly require the data type to be `torch.long`. If we don't specify this, PyTorch might throw a data type error during training.
* `X = padded_training_sequence[:, :-1]`: Slices the 2D tensor to select all rows and all columns *except* the very last one. These act as our input features (the context words).
* `y = padded_training_sequence[:, -1]`: Slices the 2D tensor to select all rows but *only* the very last column. This isolates the exact word our model needs to predict (the target).

---

### 🛠️ টেনসরে (Tensor) কনভার্ট করা এবং $X$ ও $y$ আলাদা করা (বাংলা ভার্সন)

**কোড:**

```python

# টেনসরটি চেক করা
```

**কোডের লাইন-বাই-লাইন এক্সপ্ল্যানেশন:**

* `torch.tensor(padded_sequence, ...)`: এই ফাংশনটি আমাদের তৈরি করা সাধারণ পাইথন লিস্টটিকে একটি PyTorch টেনসরে রূপান্তর করে, যা ডিপ লার্নিংয়ের ক্যালকুলেশনের জন্য বিশেষভাবে তৈরি একটি ম্যাট্রিক্স ফরম্যাট।

# **Custom Dataset and Dataloader Class**


In PyTorch, preparing data for the training loop requires two essential components:

1. **Dataset:** A class that stores all the data and knows how to retrieve a single item (one sequence and its target) at a time.
2. **DataLoader:** A wrapper around the Dataset that efficiently groups the data into "batches" (e.g., 32 sequences at once) and shuffles them so the model doesn't just memorize the order of the story.

**The Code:**

```python
# Define the Custom Dataset class
class CustomDataset(Dataset):
    def __init__(self, X, y):
        self.X = X
        self.y = y

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]
    
# Instantiate the Dataset
dataset = CustomDataset(X, y)

# Create the DataLoader
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

# Test the DataLoader
for input, output in dataloader:
    print(input, output)

```

**Line-by-Line Explanation:**

* `class CustomDataset(Dataset):`: We create a custom class that inherits from PyTorch's built-in `Dataset` class. This tells PyTorch that this class will act as our official data container.
* `def __init__(self, X, y):`: The constructor method. It receives our features ($X$) and targets ($y$) and stores them inside the class.
* `def __len__(self):`: PyTorch requires this method. It simply returns the total number of samples (rows) in our dataset so the DataLoader knows exactly how much data there is.
* `def __getitem__(self, idx):`: Another required method. When PyTorch asks for data at a specific index (`idx`), this function fetches the corresponding feature row (`self.X[idx]`) and its target word (`self.y[idx]`).
* `dataset = CustomDataset(X, y)`: We create an actual object of our custom class, passing in our sliced tensors.
* `dataloader = DataLoader(..., batch_size=32, shuffle=True)`: The DataLoader takes our `dataset` and organizes it. `batch_size=32` means it will serve 32 sequences to the model at a time. `shuffle=True` randomizes the order of the sequences in every epoch, which is critical for the model to learn the actual grammar/patterns instead of just memorizing the story line by line.
* `for input, output in dataloader:`: A test loop that grabs one batch of data at a time. Here, `input` will be a matrix of $32 \times (\text{max\_len}-1)$, and `output` will be a list of 32 target words.

### 🛠️ স্টেপ ৬: কাস্টম ডেটাসেট এবং ডেটালোডার (বাংলা ভার্সন)

PyTorch-এ মডেল ট্রেনিংয়ের জন্য ডেটাকে প্রস্তুত করতে দুটি গুরুত্বপূর্ণ টুল ব্যবহার করতে হয়:
১. **Dataset:** এটি এমন একটি ক্লাস যা সমস্ত ডেটা নিজের কাছে জমা রাখে এবং জানে কীভাবে একটি নির্দিষ্ট ইনডেক্স থেকে একটি ফিচার এবং তার টার্গেট বের করে আনতে হয়।
২. **DataLoader:** এটি Dataset-এর ওপর একটি মোড়ক (wrapper) হিসেবে কাজ করে। এর কাজ হলো ডেটাকে ছোট ছোট "ব্যাচ" বা ভাগে (যেমন: একসাথে ৩২টি সিকোয়েন্স) ভাগ করা এবং সেগুলোকে এলোমেলো (shuffle) করে মডেলের কাছে পাঠানো, যাতে মডেল গল্পের লাইনগুলো মুখস্থ না করে লজিক শিখতে পারে।

**কোড:**

```python
# Custom Dataset ক্লাস তৈরি করা

    def __init__(self, X, y):
        self.X = X
        self.y = y

    def __len__(self):

    def __getitem__(self, idx):

# Dataset-এর অবজেক্ট তৈরি করা

# DataLoader তৈরি করা
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

# DataLoader টেস্ট করা

    print(input, output)

```

**কোডের লাইন-বাই-লাইন এক্সপ্ল্যানেশন:**

* `class CustomDataset(Dataset):`: আমরা PyTorch-এর বিল্ট-ইন `Dataset` ক্লাসকে ইনহেরিট (inherit) করে একটি কাস্টম ক্লাস তৈরি করেছি। এর ফলে PyTorch বুঝতে পারে যে এই ক্লাসটি আমাদের ডেটা কন্টেইনার হিসেবে কাজ করবে।
* `def __init__(self, X, y):`: এটি ইনিশিয়ালাইজার বা কনস্ট্রাক্টর মেথড। এটি আমাদের ফিচার ($X$) এবং টার্গেট ($y$) গ্রহণ করে এবং ক্লাসের ভেতরে সেভ করে রাখে।
* `def __len__(self):`: PyTorch-এর জন্য এই মেথডটি থাকা বাধ্যতামূলক। এটি মূলত আমাদের ডেটাসেটে মোট কতগুলো স্যাম্পল বা ডেটা আছে, সেই সংখ্যাটি (length) রিটার্ন করে।
* `def __getitem__(self, idx):`: এটিও একটি বাধ্যতামূলক মেথড। PyTorch যখন কোনো নির্দিষ্ট ইনডেক্সের (`idx`) ডেটা চায়, তখন এই ফাংশনটি ঠিক সেই ইনডেক্সের ফিচার (`self.X[idx]`) এবং টার্গেট (`self.y[idx]`) রিটার্ন করে।
* `dataset = CustomDataset(X, y)`: আমরা আমাদের টেনসরগুলো ($X$ এবং $y$) দিয়ে কাস্টম ক্লাসের একটি অবজেক্ট বা ইনস্ট্যান্স তৈরি করেছি।
* `dataloader = DataLoader(..., batch_size=32, shuffle=True)`: DataLoader আমাদের `dataset`-কে প্রসেস করে। `batch_size=32`-এর মানে হলো মডেল ট্রেনিংয়ের সময় এটি একসাথে ৩২টি করে সিকোয়েন্স পাঠাবে। আর `shuffle=True`-এর মানে হলো এটি প্রতিবার ডেটার সিরিয়াল এলোমেলো করে দেবে, যা মডেলের ভালো পারফরম্যান্সের জন্য খুবই জরুরি।
* `for input, output in dataloader:`: এটি একটি টেস্ট লুপ যা DataLoader থেকে একবারে এক ব্যাচ ডেটা নিয়ে আসে। এখানে `input` হবে ৩২টি ফিচারের একটি ম্যাট্রিক্স এবং `output` হবে ৩২টি টার্গেট শব্দের একটি লিস্ট।

# **Shape of X and Y before LSTM Cell**


### 📐 Step 7: Tracking Tensor Shapes & The Embedding Layer (English Version)

Before we feed our data into the LSTM cell, it is crucial to track the shapes (dimensions) of our tensors.

#### 1. Initial Shapes of $X$ and $y$

Let's assume our dataset resulted in 555 sequences, and after padding, the maximum sequence length is 24.

* **`X.shape` $\rightarrow$ `torch.Size([555, 24])**`: This means we have 555 rows (sentences/sequences), and each row contains 24 features (words/indices).
* **`y.shape` $\rightarrow$ `torch.Size([555])**`: This means we have 555 rows, but only 1 target value (the next word to predict) for each row.

#### 2. The Embedding Transformation

An LSTM cell cannot understand simple integers (like `[1, 2, 21]`). It needs dense vectors. Therefore, we pass $X$ through an **Embedding Layer**.

Let's assume we want to represent each word using a vector of 100 numbers (Embedding Dimension = 100). When our 2D tensor $X$ passes through this layer, it gets expanded into a 3D tensor:

```text
       X (Input Tensor)
     Shape: (555, 24)
            │
            ▼
    Embedding Layer
 (100 numbers per word)
            │
            ▼
 Embedded Output Tensor
   Shape: (555, 24, 100)
            │
            ▼
        LSTM Cell

```

**Breaking down the new shape `(555, 24, 100)`:**

* **`555` (Batch Size):** Total number of sentences/sequences.
* **`24` (Sequence Length):** The number of words in each sentence.
* **`100` (Input Size/Embedding Dim):** The number of features (numbers) used to represent each individual word.

This final 3D tensor is exactly what the LSTM cell expects as its input!

### 📐 স্টেপ ৭: টেনসরের শেইপ ট্র্যাক করা এবং এমবেডিং লেয়ার (বাংলা ভার্সন)

LSTM সেলে ডেটা পাঠানোর আগে আমাদের টেনসরের শেইপ বা ডাইমেনশন ট্র্যাক করা অত্যন্ত জরুরি।

#### ১. $X$ এবং $y$-এর প্রাথমিক শেইপ

ধরে নিই, আমাদের ডেটাসেট থেকে ৫৫৫টি সিকোয়েন্স তৈরি হয়েছে এবং প্যাডিং করার পর প্রতিটি সিকোয়েন্সের সাইজ বা দৈর্ঘ্য হলো ২৪।

* **`X.shape` $\rightarrow$ `torch.Size([555, 24])**`: এর মানে হলো আমাদের ৫৫৫টি সারি (row বা বাক্য) আছে এবং প্রতিটি সারিতে ২৪টি ফিচার (শব্দ বা ইনডেক্স) আছে।
* **`y.shape` $\rightarrow$ `torch.Size([555])**`: এর মানে হলো আমাদের ৫৫৫টি সারি আছে, কিন্তু প্রতিটি সারির জন্য আউটপুট বা টার্গেট ভ্যালু মাত্র ১টি (যেই শব্দটি প্রেডিক্ট করতে হবে)।

#### ২. এমবেডিং ট্রান্সফরমেশন (Embedding Transformation)

LSTM সেল সাধারণ ইন্টিজার বা পূর্ণসংখ্যা (যেমন `[১, ২, ২১]`) বুঝতে পারে না; এর জন্য ডেন্স ভেক্টর (Dense Vector) প্রয়োজন। তাই, আমরা $X$-কে একটি **Embedding Layer**-এর ভেতর দিয়ে পাঠাব।

ধরে নিই, আমরা প্রতিটি শব্দকে ১০০টি নাম্বারের একটি ভেক্টর দিয়ে উপস্থাপন করতে চাই (Embedding Dimension = 100)। যখন আমাদের 2D টেনসর $X$ এই লেয়ারের ভেতর দিয়ে যাবে, তখন এটি একটি 3D টেনসরে রূপান্তরিত হবে:

```text
       X (ইনপুট টেনসর)
     শেইপ: (555, 24)
            │
            ▼
     Embedding Layer
  (প্রতি শব্দের জন্য ১০০ নাম্বার)
            │
            ▼
  Embedded আউটপুট টেনসর
   শেইপ: (555, 24, 100)
            │
            ▼
        LSTM Cell

```

**নতুন শেইপ `(555, 24, 100)`-এর বিস্তারিত অর্থ:**

* **`555` (Batch Size):** মোট বাক্যের বা সিকোয়েন্সের সংখ্যা।
* **`24` (Sequence Length):** প্রতিটি বাক্যে থাকা মোট শব্দের সংখ্যা।
* **`100` (Input Size/Embedding Dim):** প্রতিটি শব্দকে বোঝানোর জন্য ব্যবহৃত নাম্বারের পরিমাণ (ভেক্টর সাইজ)।

এই 3D টেনসরটিই হলো পারফেক্ট ইনপুট, যা আমরা সরাসরি আমাদের মডেলের LSTM সেলে পাঠাব!
